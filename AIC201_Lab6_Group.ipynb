# 🧪 3. Experiment #2 (Add Categorical Features)

## ⚙️ **1. Setting Up**

```python
categorical_cols = df.select_dtypes(include=['object']).columns
categorical_features = [col for col in categorical_cols if col != 'income']
```

❓ 2. Fill missing value with "unknown"

```python
for col in categorical_features:
    df[col] = df[col].fillna('Unknown')
```

## 🧹 2. Preprocess

🔤 3. One hot encode + scaler

```python
categorical_encoded = pd.get_dummies(df[categorical_features], drop_first=True)
df_exp2 = pd.concat([scaled_df, categorical_encoded], axis=1)

# Show verification results
print("New DataFrame Shape:", df_exp2.shape)
print("Missing values left in categorical features:", df[categorical_features].isnull().sum().sum())
display(df_exp2.head())
```

✂️ 4. Split the dataset

```python
X_train_exp2, X_test_exp2, y_train_exp2, y_test_exp2 = train_test_split(
    df_exp2, y_clean, test_size=0.2, random_state=42
)
```

## 🤖 3. KNN

```python
knn_exp2 = KNeighborsClassifier(n_neighbors=5, weights='uniform')
knn_exp2.fit(X_train_exp2, y_train_exp2)

y_pred_exp2 = knn_exp2.predict(X_test_exp2)

print("Accuracy (Experiment 2 - with Categoricals):", accuracy_score(y_test_exp2, y_pred_exp2))
```

### 🔁 4. Experiment #3 (Alternative) - Recursive Feature Elimination (RFE)

```python
from sklearn.feature_selection import RFE
from sklearn.linear_model import LogisticRegression
```

```python
# 1. Initialize Logistic Regression to use as the estimator for RFE
estimator = LogisticRegression(max_iter=1000, random_state=42)

# 2. Set up RFE to select 51 features (approx. 50% of our 103 original features)
rfe = RFE(estimator=estimator, n_features_to_select=51, step=5)

# Fit RFE on the training set
print("Running RFE (this may take a few moments)...")
X_train_rfe = rfe.fit_transform(X_train_exp2, y_train_exp2)
X_test_rfe = rfe.transform(X_test_exp2)

# Get names of selected columns
rfe_mask = rfe.support_
rfe_feature_names = df_exp2.columns[rfe_mask].tolist()

# 3. Train the KNN classifier on the RFE-selected features
knn_rfe = KNeighborsClassifier(n_neighbors=5, weights='uniform')
knn_rfe.fit(X_train_rfe, y_train_exp2)

# 4. Predict and evaluate
y_pred_rfe = knn_rfe.predict(X_test_rfe)

# 5. Output Results
print(f"Original number of features: {X_train_exp2.shape[1]}")
print(f"Selected number of features via RFE: {X_train_rfe.shape[1]}")
print(f"\nAccuracy (RFE Feature Selection + KNN): {accuracy_score(y_test_exp2, y_pred_rfe)}")
print("\nSample of RFE Selected Features:", rfe_feature_names[:10])
```
